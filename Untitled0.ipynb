```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import warnings
warnings.filterwarnings('ignore')

# fix seed for reproducibility
np.random.seed(42)

# load and clean data
data = np.unique(np.loadtxt('post_mansoon.txt'), axis=0)
X = data[:, [0, 1, 3]]   # Year, Rainfall, No. of Days
y = data[:, 4]            # GWL target
n = len(X)

# 70 / 15 / 15 split
tr = int(0.70 * n)
va = int(0.85 * n)
X_train, X_test = X[:tr], X[va:]
y_train, y_test = y[:tr], y[va:]

# normalize to [-1, 1]
sx = MinMaxScaler((-1, 1))
sy = MinMaxScaler((-1, 1))
X_train_s = sx.fit_transform(X_train)
X_test_s  = sx.transform(X_test)
X_all_s    = sx.transform(X)
y_train_s = sy.fit_transform(y_train.reshape(-1, 1)).ravel()

# ── nested loop: epochs x neurons = 500 experiments ─────────────
epoch_list = [50, 100, 150, 200, 250, 300, 350, 400, 450, 500]

results = []
sr = 1

print(f"\n{'Sr':>4}  {'Epochs':>6}  {'Neurons':>8}  {'R':>8}  {'MSE':>10}")
print("-" * 45)

for ep in epoch_list:
    for neurons in range(1, 51):
        np.random.seed(42)
        model = MLPRegressor(hidden_layer_sizes=(neurons,), activation='tanh',
                             solver='lbfgs', alpha=0.01,
                             max_iter=ep, random_state=42, tol=1e-12)
        model.fit(X_train_s, y_train_s)

        pred_all = sy.inverse_transform(model.predict(X_all_s).reshape(-1,1)).ravel()
        pred_te  = sy.inverse_transform(model.predict(X_test_s).reshape(-1,1)).ravel()

        R   = float(np.corrcoef(y, pred_all)[0, 1])
        MSE = float(mean_squared_error(y_test, pred_te))

        results.append({'Sr No': sr, 'Epochs': ep, 'Neurons': neurons,
                        'R Value': round(R, 4), 'MSE': round(MSE, 4)})

        print(f"{sr:>4}  {ep:>6}  {neurons:>8}  {R:>8.4f}  {MSE:>10.4f}")
        sr += 1

# ── full results table (all 500 rows) ────────────────────────────
pd.set_option('display.max_rows', 600)
pd.set_option('display.float_format', '{:.4f}'.format)

df = pd.DataFrame(results)
df = df[['Sr No', 'Epochs', 'Neurons', 'R Value', 'MSE']]

print("\n" + "=" * 50)
print("   FULL RESULTS TABLE — 500 Experiments")
print("=" * 50)
print(df.to_string(index=False))

# ── top 50 results table ──────────────────────────────────────────
top50 = df.sort_values('R Value', ascending=False).head(50).reset_index(drop=True)
top50.index = top50.index + 1
top50.index.name = 'Rank'

print("\n" + "=" * 50)
print("   TOP 50 RESULTS — Ranked by Highest R Value")
print("=" * 50)
print(top50[['Epochs', 'Neurons', 'R Value', 'MSE']].to_string())

best = top50.iloc[0]
print(f"\nBest: Epochs={int(best['Epochs'])}  Neurons={int(best['Neurons'])}"
      f"  R={best['R Value']:.4f}  MSE={best['MSE']:.4f}")

# ── lag feature model (LAG=1, R >= 0.70) ─────────────────────────
LAG = 1   # configurable: change to 2 or 3 to try more lags

y_all  = data[:, 4]
n_orig = len(data)

X_lag, y_lag = [], []
for i in range(LAG, n_orig):
    row = list(data[i, [0, 1, 3]]) + [y_all[i-l] for l in range(1, LAG+1)]
    X_lag.append(row)
    y_lag.append(y_all[i])

X_l  = np.array(X_lag)
y_l  = np.array(y_lag)
n_l  = len(X_l)

tr_l = int(0.70 * n_l)
va_l = int(0.85 * n_l)
X_train_l, X_test_l = X_l[:tr_l], X_l[va_l:]
y_train_l, y_test_l = y_l[:tr_l], y_l[va_l:]

sx_l = MinMaxScaler((-1, 1))
sy_l = MinMaxScaler((-1, 1))
X_train_ls = sx_l.fit_transform(X_train_l)
X_test_ls  = sx_l.transform(X_test_l)
X_all_ls   = sx_l.transform(X_l)
y_train_ls = sy_l.fit_transform(y_train_l.reshape(-1, 1)).ravel()

np.random.seed(42)
model_lag = MLPRegressor(hidden_layer_sizes=(30,), activation='tanh',
                         solver='lbfgs', alpha=0.01,
                         max_iter=500, random_state=42, tol=1e-12)
model_lag.fit(X_train_ls, y_train_ls)

# ── SINGLE prediction array used for BOTH R and plot ─────────────
pred_final = sy_l.inverse_transform(
                 model_lag.predict(X_all_ls).reshape(-1, 1)).ravel()

R_final   = float(np.corrcoef(y_l, pred_final)[0, 1])
MSE_final = float(mean_squared_error(y_l, pred_final))
MAE_final = float(np.mean(np.abs(y_l - pred_final)))
errors    = y_l - pred_final

print("\n" + "=" * 50)
print(f"  LAG MODEL RESULT  (LAG = {LAG})")
print("=" * 50)
print(f"  Inputs  : Year, Rainfall, NoDays + {LAG} GWL lag")
print(f"  Neurons : 30  |  Epochs: 500  |  Alpha: 0.01")
print(f"  R       : {R_final:.4f}   ← same value shown on plot")
print(f"  MSE     : {MSE_final:.4f}")
print(f"  MAE     : {MAE_final:.4f}")
print("=" * 50)

# ── regression plot — uses EXACT same pred_final as R above ───────
ax_min = float(min(y_l.min(), pred_final.min())) - 0.2
ax_max = float(max(y_l.max(), pred_final.max())) + 0.2

plt.figure(figsize=(6, 6))
plt.scatter(y_l, pred_final, color='steelblue',
            edgecolors='navy', s=50, alpha=0.8, label='Data points')
plt.plot([ax_min, ax_max], [ax_min, ax_max],
         'r--', linewidth=1.5, label='Perfect fit (1:1 line)')
plt.axis('equal')
plt.xlim(ax_min, ax_max)
plt.ylim(ax_min, ax_max)
plt.xlabel('Actual Values', fontsize=12)
plt.ylabel('Predicted Values', fontsize=12)
plt.title(f'Actual vs Predicted — GWL (m)\n'
          f'Neurons=30  Epochs=500  R={R_final:.4f}',
          fontsize=11, fontweight='bold')
plt.legend(fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('regression_plot.png', dpi=150, bbox_inches='tight')
plt.show()
print("Regression plot saved as regression_plot.png")

# ── export to Excel ───────────────────────────────────────────────
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

wb = Workbook()

# ── Sheet 1: Full 500 results ─────────────────────────────────────
ws1 = wb.active
ws1.title = "All 500 Results"

def style(cell, bold=False, bg=None, fg="000000", sz=9,
          ha="center", wrap=False):
    cell.font = Font(bold=bold, color=fg, size=sz, name="Arial")
    if bg:
        cell.fill = PatternFill("solid", start_color=bg)
    cell.alignment = Alignment(horizontal=ha, vertical="center",
                               wrap_text=wrap)
    cell.border = Border(
        top=Side(style="thin"), bottom=Side(style="thin"),
        left=Side(style="thin"), right=Side(style="thin"))

ws1.merge_cells("A1:E1")
style(ws1["A1"], bold=True, bg="1F3864", fg="FFD700", sz=13)
ws1["A1"] = "Groundwater Level Forecasting — Overall Summary Sheet (500 Experiments)"
ws1.row_dimensions[1].height = 24

ws1.merge_cells("A2:E2")
ws1["A2"].font      = Font(italic=True, color="FFFFFF", size=9, name="Arial")
ws1["A2"].fill      = PatternFill("solid", start_color="2E75B6")
ws1["A2"].alignment = Alignment(horizontal="center", vertical="center")
ws1["A2"] = "Epochs: 50 to 500  |  Neurons: 1 to 50  |  Activation: tansig  |  Solver: lbfgs  |  Seed: 42  |  Split: 70/15/15"
ws1.row_dimensions[2].height = 15

for i, w in enumerate([8, 10, 10, 12, 12], 1):
    ws1.column_dimensions[get_column_letter(i)].width = w

headers = ["Sr No", "Epochs", "Neurons", "R Value", "MSE"]
for ci, h in enumerate(headers, 1):
    c = ws1.cell(row=3, column=ci, value=h)
    style(c, bold=True, bg="17375E", fg="FFFFFF", sz=9)
ws1.row_dimensions[3].height = 18

best_R_val = df['R Value'].max()
for ri, row in df.iterrows():
    excel_row = ri + 4
    is_best   = (row['R Value'] == best_R_val)
    is_top    = (row['R Value'] >= 0.63)
    is_good   = (row['R Value'] >= 0.58)
    bg = ("C6EFCE" if is_best else
          "E2EFDA" if is_top else
          "EBF3FB" if is_good else
          ("F5F5F5" if ri % 2 == 0 else "FFFFFF"))
    fg = "1E4620" if is_best else "000000"
    vals = [row['Sr No'], row['Epochs'], row['Neurons'],
            row['R Value'], row['MSE']]
    for ci, val in enumerate(vals, 1):
        c = ws1.cell(row=excel_row, column=ci, value=val)
        style(c, bold=is_best, bg=bg, fg=fg, sz=9)
        if ci in (4, 5):
            c.number_format = "0.0000"
    ws1.row_dimensions[excel_row].height = 13
ws1.freeze_panes = "A4"

# ── Sheet 2: Top 50 results ───────────────────────────────────────
ws2 = wb.create_sheet("Top 50 Results")
ws2.merge_cells("A1:E1")
ws2["A1"] = "Top 50 Results — Ranked by Highest R Value"
ws2["A1"].font      = Font(bold=True, color="FFD700", size=13, name="Arial")
ws2["A1"].fill      = PatternFill("solid", start_color="1F3864")
ws2["A1"].alignment = Alignment(horizontal="center", vertical="center")
ws2.row_dimensions[1].height = 24

for i, w in enumerate([8, 10, 10, 12, 12], 1):
    ws2.column_dimensions[get_column_letter(i)].width = w

for ci, h in enumerate(["Rank", "Epochs", "Neurons", "R Value", "MSE"], 1):
    c = ws2.cell(row=2, column=ci, value=h)
    style(c, bold=True, bg="17375E", fg="FFFFFF", sz=9)
ws2.row_dimensions[2].height = 18

top50_reset = df.sort_values('R Value', ascending=False).head(50).reset_index(drop=True)
for ri, row in top50_reset.iterrows():
    rank = ri + 1; excel_row = ri + 3
    is_first = (rank == 1); is_top3 = (rank <= 3)
    bg = ("C6EFCE" if is_first else
          "E2EFDA" if is_top3 else
          ("EBF3FB" if rank <= 10 else
           ("F5F5F5" if ri % 2 == 0 else "FFFFFF")))
    fg = "1E4620" if is_first else "000000"
    for ci, val in enumerate(
            [rank, row['Epochs'], row['Neurons'], row['R Value'], row['MSE']], 1):
        c = ws2.cell(row=excel_row, column=ci, value=val)
        style(c, bold=is_first, bg=bg, fg=fg, sz=9)
        if ci in (4, 5): c.number_format = "0.0000"
    ws2.row_dimensions[excel_row].height = 13
ws2.freeze_panes = "A3"

# ── Sheet 3: Best config ──────────────────────────────────────────
ws3 = wb.create_sheet("Best Configuration")
ws3.merge_cells("A1:B1")
ws3["A1"] = "Best Configuration Found"
ws3["A1"].font      = Font(bold=True, color="FFD700", size=13, name="Arial")
ws3["A1"].fill      = PatternFill("solid", start_color="1F3864")
ws3["A1"].alignment = Alignment(horizontal="center", vertical="center")
ws3.column_dimensions["A"].width = 25
ws3.column_dimensions["B"].width = 30
ws3.row_dimensions[1].height = 24

best_info = [
    ("Epochs",            str(int(best['Epochs']))),
    ("Hidden Neurons",    str(int(best['Neurons']))),
    ("Hidden Activation", "tansig"),
    ("Output Activation", "linear"),
    ("Solver",            "lbfgs"),
    ("Alpha (L2 reg)",    "0.01"),
    ("Split",             "70% / 15% / 15%"),
    ("Random Seed",       "42"),
    ("R Value",           f"{best['R Value']:.4f}"),
    ("MSE",               f"{best['MSE']:.4f} m\u00b2"),
]
for ri, (label, val) in enumerate(best_info, 2):
    cl = ws3.cell(row=ri, column=1, value=label)
    cv = ws3.cell(row=ri, column=2, value=val)
    is_result = label in ("R Value", "MSE")
    style(cl, bold=True, bg="BDD7EE", fg="1F3864", sz=10, ha="left")
    style(cv, bold=is_result, bg="C6EFCE" if is_result else "FFFFFF",
          fg="1E4620" if is_result else "000000", sz=10, ha="left")
    ws3.row_dimensions[ri].height = 16

# ── Sheet 4: Input + Actual + Predicted + Error ───────────────────
ws4 = wb.create_sheet("Predictions")
ws4.merge_cells("A1:G1")
ws4["A1"] = f"Lag Model Predictions  |  R = {R_final:.4f}  |  MSE = {MSE_final:.4f}  |  MAE = {MAE_final:.4f}"
ws4["A1"].font      = Font(bold=True, color="FFD700", size=11, name="Arial")
ws4["A1"].fill      = PatternFill("solid", start_color="1F3864")
ws4["A1"].alignment = Alignment(horizontal="center", vertical="center")
ws4.row_dimensions[1].height = 20

pred_headers = ["Year", "Rainfall (mm)", "No. of Days",
                "GWL Lag-1", "Actual GWL", "Predicted GWL", "Error"]
pred_widths  = [8, 14, 12, 12, 13, 15, 10]
for ci, (h, w) in enumerate(zip(pred_headers, pred_widths), 1):
    ws4.column_dimensions[get_column_letter(ci)].width = w
    c = ws4.cell(row=2, column=ci, value=h)
    style(c, bold=True, bg="17375E", fg="FFFFFF", sz=9)
ws4.row_dimensions[2].height = 18

for ri in range(len(X_l)):
    excel_row = ri + 3
    alt_bg = "F5F5F5" if ri % 2 == 0 else "FFFFFF"
    row_vals = [
        int(X_l[ri, 0]),          # Year
        round(X_l[ri, 1], 2),     # Rainfall
        int(X_l[ri, 2]),           # NoDays
        round(X_l[ri, 3], 4),     # GWL lag-1
        round(y_l[ri], 4),         # Actual
        round(pred_final[ri], 4),  # Predicted
        round(errors[ri], 4),      # Error
    ]
    for ci, val in enumerate(row_vals, 1):
        c = ws4.cell(row=excel_row, column=ci, value=val)
        style(c, bg=alt_bg, sz=9)
        if ci in (4, 5, 6, 7): c.number_format = "0.0000"
    ws4.row_dimensions[excel_row].height = 13

# global metrics row at the bottom
last_row = len(X_l) + 4
ws4.merge_cells(f"A{last_row}:D{last_row}")
c = ws4.cell(row=last_row, column=1,
             value=f"Global Metrics  →  R = {R_final:.4f}   MSE = {MSE_final:.4f}   MAE = {MAE_final:.4f}")
style(c, bold=True, bg="1F3864", fg="FFD700", sz=9, ha="left")
ws4.row_dimensions[last_row].height = 16
ws4.freeze_panes = "A3"

wb.save("Overall_Summary_Sheet.xlsx")
print("\nExcel file saved: Overall_Summary_Sheet.xlsx")

# ── download ──────────────────────────────────────────────────────
from google.colab import files
files.download('Overall_Summary_Sheet.xlsx')
files.download('regression_plot.png')
```
